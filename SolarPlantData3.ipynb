```python
plantgen['DATE_TIME'] = pd.to_datetime(plantgen['DATE_TIME'], errors='coerce')
plantweather['DATE_TIME'] = pd.to_datetime(plantweather['DATE_TIME'], errors='coerce')

plantgen['DATE'] = plantgen['DATE_TIME'].dt.date

plantgen = plantgen.sort_values(by=['SOURCE_KEY', 'DATE_TIME'])

plantgen['INTERVAL_ENERGY'] = plantgen['AC_POWER'] * 0.25

plantgen['TRUE_DAILY_YIELD'] = plantgen.groupby(['SOURCE_KEY','DATE'])['INTERVAL_ENERGY'].cumsum()

plantgen_agg = plantgen.groupby('DATE_TIME').agg({'PLANT_ID': 'first', 'DC_POWER': 'sum', 
                    'AC_POWER': 'sum', 'TRUE_DAILY_YIELD': 'sum'}).reset_index()
plantgen_agg.iloc[1:30]
```

```python
plantweather.drop(columns=['SOURCE_KEY'], inplace=True)
plantweather.head()
```

```python
plantdata = pd.merge(plantgen_agg, plantweather, on=['DATE_TIME', 'PLANT_ID'], how='left')
plantdata.head()
```

```python
plantdata['DCE'] = (plantdata['DC_POWER'] / plantdata['DC_POWER'].max()).fillna(0)
plantdata['ACE'] = (plantdata['AC_POWER'] / plantdata['AC_POWER'].max()).fillna(0)
plantdata['IE'] = (plantdata['AC_POWER'] / plantdata['DC_POWER'].max()).fillna(0)
plantdata['YI'] = (plantdata['DC_POWER'] / plantdata['IRRADIATION']).fillna(0)
```

```python
plantdata.head(10)
```

```python
plantdata = plantdata.rename(columns={'DCE': 'DC_EFFICIENCY', 'ACE': 'AC_EFFICIENCY', 'IE': 'INVERTOR_EFFICIENCY', 
                                      'YI': 'YIELD_INDEX'})
plantdata = plantdata[['DATE_TIME', 'PLANT_ID', 'IRRADIATION', 'AMBIENT_TEMPERATURE', 'DC_EFFICIENCY', 'AC_EFFICIENCY', 'INVERTOR_EFFICIENCY', 'YIELD_INDEX', 'TRUE_DAILY_YIELD']]
plantdata.iloc[20:30]
```

```python
plantdata.to_csv('SOLAR_PLANT_DATA(GENERATION_AND_WEATHER).csv', index=False)
```
