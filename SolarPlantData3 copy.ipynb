```python
plantgen['DATE_TIME'] = pd.to_datetime(plantgen['DATE_TIME'], errors='coerce')
plantweather['DATE_TIME'] = pd.to_datetime(plantweather['DATE_TIME'], errors='coerce')

plantgen['DATE'] = plantgen['DATE_TIME'].dt.date

plantgen = plantgen.sort_values(by=['SOURCE_KEY', 'DATE_TIME'])

plantgen['INTERVAL_ENERGY'] = plantgen['AC_POWER'] * 0.25

plantgen['DAILY_YIELD'] = plantgen.groupby(['SOURCE_KEY','DATE'])['INTERVAL_ENERGY'].cumsum()

plantgen['TOTAL_YIELD'] = plantgen.groupby(['SOURCE_KEY'])['INTERVAL_ENERGY'].cumsum()

plantgen_agg = plantgen.groupby('DATE_TIME').agg({'PLANT_ID': 'first', 'DC_POWER': 'sum', 
                    'AC_POWER': 'sum', 'DAILY_YIELD': 'sum', 'TOTAL_YIELD': 'sum'}).reset_index()
plantgen_agg.iloc[200:220]
```

```python
plantweather.drop(columns=['SOURCE_KEY'], inplace=True)
plantweather.head()
```

```python
plantdata = pd.merge(plantgen_agg, plantweather, on=['DATE_TIME', 'PLANT_ID'], how='left')
plantdata.head()
```

```python
plantdata['DCE'] = (plantdata['DC_POWER'] / plantdata['DC_POWER'].max()).fillna(0)
plantdata['ACE'] = (plantdata['AC_POWER'] / plantdata['AC_POWER'].max()).fillna(0)
plantdata['IE'] = (plantdata['AC_POWER'] / plantdata['DC_POWER'].max()).fillna(0)
plantdata['YI'] = (plantdata['DC_POWER'] / plantdata['IRRADIATION']).fillna(0)
plantdata['IRRADIATION'] = plantdata['IRRADIATION'] * 1000
```

```python
plantdata.head(10)
```

```python
plantdata = plantdata.rename(columns={'DC_POWER':'DC_POWER(kW)', 'AC_POWER':'AC_POWER(kW)', 'DCE': 'DC_EFFICIENCY', 'ACE': 'AC_EFFICIENCY', 'IE': 'INVERTOR_EFFICIENCY', 
                                      'YI': 'EFFECTIVE_AREA(m²)', 'DAILY_YIELD': 'DAILY_YIELD(kWh)', 'TOTAL_YIELD': 'TOTAL_YIELD(kWh)', 
                                      'AMBIENT_TEMPERATURE': 'AMBIENT_TEMPERATURE(°C)', 'IRRADIATION': 'IRRADIATION(W/m²)', 'MODULE_TEMPERATURE': 'MODULE_TEMPERATURE(°C)'})
plantdata = plantdata[['DATE_TIME', 'PLANT_ID', 'IRRADIATION(W/m²)', 'AMBIENT_TEMPERATURE(°C)', 'MODULE_TEMPERATURE(°C)', 'DC_POWER(kW)','AC_POWER(kW)', 
                        'DC_EFFICIENCY', 'AC_EFFICIENCY', 'INVERTOR_EFFICIENCY', 'EFFECTIVE_AREA(m²)', 'DAILY_YIELD(kWh)', 'TOTAL_YIELD(kWh)']]
plantdata.iloc[20:30]
```

```python
plantdata.to_csv('SOLAR_PLANT_DATA(GENERATION_AND_WEATHER).csv', index=False)
```
